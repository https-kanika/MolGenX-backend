```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

def process_affinity_type(df, affinity_type, threshold=0.6):
    """
    Process one affinity type: log transform, scale, and filter by threshold
    
    Parameters:
    -----------
    df : pandas DataFrame
        Input DataFrame containing affinity data
    affinity_type : str
        The affinity type to process ('Ki', 'IC50', etc.)
    threshold : float
        Minimum scaled affinity value to keep
        
    Returns:
    --------
    pandas DataFrame
        Processed DataFrame with scaled affinities for this type
    """
    print(f"\nProcessing {affinity_type} values...")
    
    # Filter to just this affinity type
    df_type = df[df['standard_type'] == affinity_type].copy()
    original_count = len(df_type)
    print(f"Found {original_count} {affinity_type} measurements")
    
    if original_count == 0:
        return pd.DataFrame()  # No data for this type
    
    # Remove invalid values (zero or negative)
    df_type = df_type[df_type['standard_value'] > 0]
    filtered_count = len(df_type)
    print(f"After removing zero/negative values: {filtered_count} measurements")
    
    if filtered_count == 0:
        return pd.DataFrame()  # No valid data for this type
    
    # Log transform the values
    df_type['log_affinity'] = np.log(df_type['standard_value'])
    
    # Scale the values between 0 and 1
    scaler = MinMaxScaler()
    df_type['scaled_affinity'] = scaler.fit_transform(df_type[['log_affinity']])
    
    # Show original value ranges
    min_val = df_type['standard_value'].min()
    max_val = df_type['standard_value'].max()
    print(f"Original {affinity_type} range: {min_val:.2e} to {max_val:.2e}")
    
    # Show log transformed ranges
    min_log = df_type['log_affinity'].min()
    max_log = df_type['log_affinity'].max()
    print(f"Log transformed range: {min_log:.2f} to {max_log:.2f}")
    
    # Show scaling transformation parameters
    print(f"MinMaxScaler parameters: min={scaler.data_min_[0]:.2f}, max={scaler.data_max_[0]:.2f}")
    
    # Filter by threshold
    df_filtered = df_type[df_type['scaled_affinity'] >= threshold].copy()
    final_count = len(df_filtered)
    print(f"After filtering (scaled_affinity >= {threshold}): {final_count} measurements")
    print(f"Removed {filtered_count - final_count} measurements ({(filtered_count - final_count) / filtered_count * 100:.1f}% of valid data)")
    
    return df_filtered

def process_all_affinity_types(df, threshold=0.6):
    """
    Process all affinity types in the dataset
    """
    # Get all unique affinity types
    affinity_types = df['standard_type'].unique()
    print(f"Found {len(affinity_types)} affinity types: {affinity_types}")
    
    # Process each type
    processed_dfs = []
    for affinity_type in affinity_types:
        df_processed = process_affinity_type(df, affinity_type, threshold)
        processed_dfs.append(df_processed)
    
    # Combine all processed dataframes
    df_combined = pd.concat(processed_dfs, ignore_index=True)
    
    print("\n=== Summary ===")
    print(f"Original dataset size: {len(df)}")
    print(f"Final dataset size: {len(df_combined)}")
    print(f"Retained {len(df_combined) / len(df) * 100:.1f}% of the original data")
    
    return df_combined

# Usage example
# processed_df = process_all_affinity_types(df)
```
